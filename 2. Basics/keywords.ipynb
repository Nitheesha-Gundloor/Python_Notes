1. Keywords

Keywords are reserved words in Python that have a special meaning. They cannot be used as variable names.

keywords : False, None, True, and, as, assert, async, await, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lambda, nonlocal, not, or, pass, raise, return, try, while, with, yield

2. Identifiers

Identifiers are names given to variables, functions, classes, objects, etc.

```python
name = "John"       # name → identifier
age = 20            # age → identifier
print(name, age)

#Simple rule: Identifier = Name given to something in a program.
```

3. Functions

A function is a reusable block of code that performs a specific task.

```python
def greet():
    print("Hello")

greet()
```

4. Classes

A class is a blueprint/template used to create objects.

```python
class Student:
    name = "John"

#Here, Student is a class.
```

5. Objects

An object is an instance of a class. It represents a specific entity created from the class.

```python
class Student:
    name = "John"

s1 = Student()
print(s1.name)
#Here, s1 is an object of the Student class.
```

6. del Keyword

The del keyword is used to delete a variable, object, item from a list, etc.

```python
age = 20
print(age)

del age
print(age)                                              #age is deleted so NameError occurs
```
