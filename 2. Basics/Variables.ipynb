1. Rules for Creating Variables in Python
Must start with a letter or underscore (_).

```python
name = "John"
_age = 20
print(name,_age)
```

2. Cannot start with a number.

```python
1name = "John"   # ❌ Invalid
name1 = "John"  # ✅ Valid
print(name1,1name)
```

3. Can contain letters, numbers, and underscores.

```python
student_name = "John"
age2 = 21
print(student_name,age2)
```

4. Cannot contain spaces or special characters.

```python
student name = "John"   # ❌
student_name = "John"   # ✅
print(student name,student_name)
```

5. Python keywords cannot be used as variable names.

```python
if = 10       # ❌
value = 10    # ✅
```

6. Variable names are case-sensitive.

```python
age = 20
Age = 30
# Both are different variables
print(age,Age)
```
