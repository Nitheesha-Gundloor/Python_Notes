!!! INSTALLATION OF PYTHON !!!
1. Download Python: Go to the official website python.org and click Download Python.
2. Run Installer: Open the downloaded .exe file to start the installation.
3. Enable PATH: Check “Add Python.exe to PATH” at the bottom of the installer.
4. Install Python: Click Install Now and wait for the installation to complete.
5. Verify Installation: Open Command Prompt and type: python --version

WHAT IS PYTHON?

Python is a high-level, interpreted, general-purpose programming language used to create software, automate tasks, analyze data, build websites, and develop AI/ML applications.

WHY WE SHOULD USE PYTHON? 

Easy to learn, Huge library, Built-In Functions, Automation, Web Development, Data Science, AI & Machine Learning, Large Community

Applications : Python is applied in Web Development, Data Science, AI/ML, Automation, Game Development, Web Scraping, Scientific Computing, Cybersecurity, and RPA.

PROCEDURAL PROGRAMMING
-- Focuses on functions and step-by-step instructions.
-- Data and functions are generally separate.
-- Suitable for small and simple programs.

OOP (Object-Oriented Programming)
-- Focuses on objects and classes.
-- Combines data and functions into objects.
--Suitable for large and complex programs.
