# Strings in PythonA **string** is just text — anything wrapped in quotes (`'single'`, `"double"`, or `'''triple'''`).Strings are one of the most used data types, so knowing how to slice, search, and clean them up will save you a lot of time.

## Indexing and SlicingStrings are sequences, so you can grab characters by position. Index starts at `0`.Negative index counts from the end.

```python
text = "Python"print(text[0])     # first characterprint(text[-1])    # last characterprint(text[0:3])   # slice: first 3 charactersprint(text[::-1])  # reverse the whole string
```

## Strings are ImmutableOnce created, you can't change a string in place. Any "change" actually creates a new string.

```python
word = "cat"new_word = word.replace("c", "b")print(word)      # unchangedprint(new_word)  # new string created
```

## Common String MethodsHere are the ones you'll use constantly:| Method | What it does ||---|---|| `.upper()` / `.lower()` | change case || `.strip()` | remove extra spaces from both ends || `.split()` | break string into a list || `.join()` | combine a list into a string || `.replace()` | swap out text || `.find()` | get index of a substring || `.startswith()` / `.endswith()` | check beginning/end |

```python
s = "  Hello World  "print(s.upper())print(s.lower())print(s.strip())print(s.strip().split())        # split by spaces -> listprint("-".join(["a", "b", "c"])) # join list into a stringprint(s.replace("World", "Python"))print(s.find("World"))print(s.strip().startswith("Hello"))
```

## f-strings (formatting made easy)The cleanest way to insert variables into text.

```python
name = "Alice"age = 25print(f"{name} is {age} years old")print(f"Next year, {name} will be {age + 1}")
```

## Quick Recap- Strings are sequences of characters, indexed from `0`.- They're immutable — methods return *new* strings, they don't change the original.- `.upper()`, `.lower()`, `.strip()`, `.split()`, `.join()`, `.replace()` cover most everyday needs.- Use f-strings for clean, readable formatting.
