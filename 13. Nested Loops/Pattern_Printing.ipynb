# Pattern Printing

6 classic patterns using nested loops.

### 1. Square

```python
n = 4
for i in range(n):
    print("* " * n)
```

### 2. Right Triangle

```python
n = 5
for i in range(1, n + 1):
    print("* " * i)
```

### 3. Inverted Right Triangle

```python
n = 5
for i in range(n, 0, -1):
    print("* " * i)
```

### 4. Pyramid

```python
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
```

### 5. Number Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

### 6. Diamond

```python
n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)
```

7. Plus Shape


```python
n=5
for r in range(n):
    for c in range(n):
        if r == n//2 or c == n//2:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

8. Number Matrix

```python
n=3
i=1
for r in range(n):
    for c in range(n):
        print(i,end=" ")
        i+=1
    print()
```

9. 10 10 bits 

```python
n=4
for r in range(n):
    for c in range(4):
        if (r+c) % 2 == 0:
            print("1", end=" ")
        else:
            print("0",end=" ")
    print()
```
