# Finding Largest Elements Using LoopsSame goal — find top elements — solved with different loop approaches.

### Find the first (largest) element — using a for loop

```python
nums = [10, 45, 3, 89, 22]largest = nums[0]for n in nums:    if n > largest:        largest = nprint("Largest:", largest)
```

### Find the second largest — using a while loop

```python
nums = [10, 45, 3, 89, 22]largest = second = float('-inf')i = 0while i < len(nums):    if nums[i] > largest:        second = largest        largest = nums[i]    elif nums[i] > second:        second = nums[i]    i += 1print("Second largest:", second)
```

### Find the third largest — using sorting

```python
nums = [10, 45, 3, 89, 22]unique_sorted = sorted(set(nums), reverse=True)third_largest = unique_sorted[2]print("Third largest:", third_largest)
```
