# While LoopA `while` loop keeps running as long as its condition stays `True`. It's used when you don't know exactly how many times you'll need to repeat — you just know the stopping condition.

### Problem 1: Sum of first N natural numbers

```python
n = 5total = 0i = 1while i <= n:    total += i    i += 1print("Sum:", total)
```

### Problem 2: Reverse a number

```python
num = 1234reversed_num = 0while num > 0:    digit = num % 10    reversed_num = reversed_num * 10 + digit    num //= 10print("Reversed number:", reversed_num)
```
