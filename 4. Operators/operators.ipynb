ARITHMETIC OPERATIONS:

 (+ - * / // % **) perform basic math like addition, subtraction, division, and exponentiation on numbers.

```python
a = 10
b = 3
print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.3333333333333335
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

Comparison operators:
 (== != > < >= <=) compare two values and return True or False.

```python
a = 10
b = 20
print(a == b)  # False
print(a != b)  # True
print(a > b)   # False
print(a < b)   # True
print(a >= b)  # False
print(a <= b)  # True
```

Logical operators: (and, or, not) combine or invert boolean conditions.

```python
a = True
b = False
print(a and b)  # False
print(a or b)   # True
print(not a) 
```

Assignment operators:
 (= += -= *= /= //= %= **=) assign a value to a variable, optionally combined with an arithmetic operation.

```python
x = 10
x += 5   # x = 15
print(x)

x -= 3   # x = 12
print(x)

x *= 2   # x = 24
print(x)

x //= 5  # x = 4
print(x)
```

Bitwise operators : 
(& | ^ ~ << >>) manipulate numbers at the level of their individual binary bits.

```python
a = 10   # binary: 1010
b = 4    # binary: 0100
print(a & b)   # 0
print(a | b)   # 14
print(a ^ b)   # 14
print(~a)      # -11
print(a << 2)  # 40
print(a >> 2)  # 2
```

Membership operators:
 (in, not in) check whether a value exists within a sequence like a list, tuple, or string.

```python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)      # True
print("mango" in fruits)       # False
print("mango" not in fruits)   # True
```

Identity operators:
 (is, is not) check whether two variables refer to the exact same object in memory, not just equal values.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a is c)      # True  (c points to the same object as a)
print(a is b)      # False (different objects, same values)
print(a == b)      # True  (values are equal)
print(a is not b)  # True
```
