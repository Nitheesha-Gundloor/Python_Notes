# Sets and Dictionaries- A **set** is an unordered collection of *unique* items — no duplicates allowed.- A **dictionary** stores data as **key: value** pairs — great for looking things up by name instead of position.

## Sets

```python
numbers = {1, 2, 3, 3, 2}   # duplicates get dropped automaticallyprint(numbers)empty_set = set()   # note: {} creates a dict, not a setprint(type(empty_set))
```

### Common Set Methods

```python
fruits = {"apple", "banana", "cherry"}fruits.add("orange")print(fruits)fruits.remove("banana")print(fruits)a = {1, 2, 3}b = {2, 3, 4}print(a.union(b))         # all unique elements from bothprint(a.intersection(b))  # elements common to bothprint(a.difference(b))    # elements only in a
```

Sets are perfect for quickly removing duplicates or checking membership.

```python
nums_list = [1, 2, 2, 3, 3, 3, 4]unique_nums = list(set(nums_list))print(unique_nums)print(3 in a)   # fast membership check
```

## Dictionaries

```python
person = {    "name": "Alice",    "age": 25,    "city": "New York"}print(person)print(person["name"])      # access by keyprint(person.get("age"))   # safer access, no error if missing
```

### Common Dictionary Methods| Method | What it does ||---|---|| `.keys()` | all the keys || `.values()` | all the values || `.items()` | key-value pairs || `.get()` | safe access (no error if key missing) || `.update()` | add/change multiple key-values || `.pop()` | remove a key (returns its value) |

```python
print(person.keys())print(person.values())print(person.items())person.update({"age": 26, "job": "Engineer"})print(person)removed = person.pop("city")print("Removed:", removed)print(person)
```

Dictionaries are also **mutable** — you can add or change entries anytime.

```python
person["age"] = 27          # update existing keyperson["country"] = "USA"   # add a new keyprint(person)
```

### Looping Through a Dictionary

```python
for key, value in person.items():    print(key, "->", value)
```

### Quick Bonus: Dictionary ComprehensionA one-liner way to build a dictionary.

```python
squares = {x: x**2 for x in range(1, 6)}print(squares)
```

## Quick Recap- Sets store unique, unordered items — great for de-duplication and membership checks.- Dictionaries store key-value pairs — great for structured, named data.- Both are mutable and support handy built-in methods for adding, removing, and combining data.
