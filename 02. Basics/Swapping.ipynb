What is Swapping in Python?

Swapping means exchanging the values of two variables.

For Examples:
Before swap  ->> a = 10, b = 20
After swap   ->> a=20, b=10

1. Swapping using a Temporary Variable

This is the traditional method used in many programming languages.

How it works:
temp = a    → temp = 10
a = b       → a = 20
b = temp    → b = 10

Here, temp temporarily stores the value of a.

```python
a = 10
b = 20

temp = a
a = b
b = temp

print(a)
print(b)
```

2. Swapping Without a Temporary Variable – Arithmetic

We can use + and - to swap numbers.

Step-by-step:
a = 10, b = 20

a = a + b  → 30
b = a - b  → 10
a = a - b  → 20

⚠️ This method is mainly useful for understanding the concept. In Python, the next method is preferred.

```python
a = 10
b = 20
a = a + b
b = a - b
a = a - b
print(a)
print(b)
```

3. Swapping Using Multiplication and Division
This has limitations, such as problems when one value is 0, so it is not recommended in practical Python programming.

```python
a = 10
b = 20
a = a * b
b = a // b
a = a // b
print(a)
print(b)
```

4. Swapping Using Python's Multiple Assignment ⭐

This is the Pythonic and recommended method.

How it works:
a, b = b, a

means:
a gets the value of b
b gets the value of a

No temporary variable is required.

```python
a = 10
b = 20
a, b = b, a
print(a)
print(b)
```

5. Swapping Using XOR

For integer values, swapping can also be done using the XOR (^) operator.

This is a bitwise technique and is mainly useful for understanding bitwise operations. It is generally unnecessary in normal Python code.

```python
a = 10
b = 20
a = a ^ b
b = a ^ b
a = a ^ b
print(a)
print(b)
```
