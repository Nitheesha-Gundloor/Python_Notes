# Control Statements and LoopsControl statements decide the flow of a program — which code runs, how many times, and when to stop. Loops are the main tool for repeating work.

## 3 Types of Loops in Python1. **for loop** — repeats over a known sequence (list, string, range, etc.)2. **while loop** — repeats as long as a condition stays true3. **nested loop** — a loop inside another loop, useful for grids/patterns

```python
# for loopfor i in range(3):    print("for loop:", i)# while loopn = 0while n < 3:    print("while loop:", n)    n += 1# nested loopfor i in range(2):    for j in range(2):        print("nested:", i, j)
```

## What is `range()`?`range()` generates a sequence of numbers — great for looping a fixed number of times without creating an actual list.

```python
print(list(range(5)))        # 0 to 4print(list(range(2, 8)))     # 2 to 7print(list(range(0, 10, 2))) # step of 2
```

## for loop on a list

```python
fruits = ["apple", "banana", "cherry"]for fruit in fruits:    print(fruit)
```

## Jump Statements- `break` — stops the loop completely- `continue` — skips the current step, moves to the next- `pass` — does nothing, just a placeholder

```python
for i in range(5):    if i == 3:        break          # stop when i is 3    print("break demo:", i)for i in range(5):    if i == 3:        continue       # skip 3, keep going    print("continue demo:", i)for i in range(3):    if i == 1:        pass            # placeholder, does nothing    print("pass demo:", i)
```

## assert keyword`assert` checks a condition and stops the program with an error if it's False. Useful for catching bugs early.

```python
age = 20assert age >= 18, "Age must be 18 or above"print("Assertion passed, age is valid")
```
