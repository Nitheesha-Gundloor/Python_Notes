What is a data type?
A data type describes the kind of object/value Python is working with and determines the operations that are meaningful for that object.

Category               	   Common Python types           	Example
1. Numeric	 :                int, float, complex	         10, 3.14, 2+3j
2. Boolean   :                  bool	                    True, False
3. Text	     :                  str	                       "Python"
4. Sequence  :          	list, tuple, range              [1,2], (1,2), range(5)
5. Set	     :                  set, frozenset	            {1,2}, frozenset({1,2})
6. Mapping	 :                 dict	                        {"a": 1}
7. Binary	 :           bytes, bytearray, memoryview	     b'abc'
8. None	     :                 NoneType                      None


1. List
 
 * Mutable Datatype
Definition: A list is an ordered, changeable (mutable) collection that allows duplicate values. 

```python
fruits = ["apple", "banana", "apple", "mango"]
print(fruits)
```

2. Tuple
 * Immutable Datatype
Definition: A tuple is an ordered, unchangeable (immutable) collection that allows duplicate values.

```python
numbers = (10, 20, 30, 10)
print(numbers)
```

3. Set

*MUTABLE

Definition: A set is an unordered, mutable collection of unique values that does not allow duplicates.

```python
numbers = {10, 20, 30, 40}
print(numbers)
```

4. Dictionary

* MUTABLE

Definition: A dictionary is a mutable collection of key-value pairs where keys must be unique.

```python
student = {"name": "Rahul", "age": 21, "branch": "CSE"}
print(student)
```

5. Complex

* Immutable

Definition: A complex number consists of a real part and an imaginary part, represented using j.

```python
z = 3 + 4j
```

* List → [] → Mutable
* Tuple → () → Immutable
* Set → {} → Mutable
* Dictionary → {key: value} → Mutable
* Complex → 3 + 4j → Immutable

!!! TYPE CASTING

Definition: Type casting is the process of converting a value from one data type to another data type.

Types of Type Casting

1. Implicit Type Casting

Definition: Python automatically converts one data type into another without the programmer explicitly converting it.
Type: Automatic conversion

2. Explicit Type Casting

Definition: The programmer manually converts a value from one data type to another using built-in functions.
Type: Manual conversion

Implicit Type Casting → Done automatically by Python
Explicit Type Casting  → Done manually by the programmer

```python
x = float("25.5")    # changes string to float
print(x)

x = str(100)
print(x)              # integer to string
print(type(x))

x = bool(10)           # integer to bool
print(x)

x = list("ABC")       # string to list
print(x)

x = tuple([1, 2, 3])  # list to tuple
print(x)

x = set([1, 2, 2, 3])  # list to set
print(x)

#EXAMPLE:
# Explicit
x = int("10")

# Implicit
x = 10 + 2.5
```
