# Conditional Statements in PythonConditional statements let your program **make decisions**. Depending on whether something is `True` or `False`, Python chooses which block of code to run.There are 4 main ones you'll use all the time:

## 1. `if`The simplest one. It checks a condition, and if that condition is `True`, it runs the code inside it. If it's `False`, that block is just skipped — nothing happens.

## 2. `if...else`This adds a backup plan. If the condition is `True`, the `if` block runs. If it's `False`, Python runs the `else` block instead. One of the two always runs — never both.

## 3. `if...elif...else`Used when you have more than two possibilities. Python checks each condition top to bottom, and runs the first one that turns out `True`. If none of them are `True`, it falls back to `else`. Only one block ever runs, even if multiple conditions could technically be true.

## 4. Nested `if`This is simply an `if` statement placed inside another `if` (or `else`) block. It's useful when a decision only makes sense *after* another condition has already been satisfied — like checking a second detail only if the first one passes.

## Quick Recap- `if` → run something only when a condition is true.- `if...else` → pick between two paths.- `if...elif...else` → pick one path out of many.- Nested `if` → a decision inside another decision.
