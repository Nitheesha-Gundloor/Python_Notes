# Name & Alphabet Pattern Programs

## Name Patterns using "NItheesha"

### 1. Growing name triangle

```python
name = "NItheesha"

for i in range(1, len(name) + 1):
    print(name[:i])
```

### 2. Name repeated as a pyramid

```python
name = "NItheesha"
n = len(name)

for i in range(1, n + 1):
    print(" " * (n - i) + name[:i])
```

## Alphabet / ASCII Patterns

### 1. Same letter repeated per row (increasing letter)

```python
n = 5

for i in range(n):
    letter = chr(65 + i)
    print(letter * (i + 1))
```

### 2. Different letters in each row

```python
n = 5

for i in range(1, n + 1):
    row = ""
    for j in range(i):
        row += chr(65 + j) + " "
    print(row)
```

### 3. Alphabet pyramid (centered)

```python
n = 5

for i in range(1, n + 1):
    letters = "".join(chr(65 + j) for j in range(i))
    print(" " * (n - i) + letters)
```

### 4. Reverse alphabet triangle

```python
n = 5

for i in range(n, 0, -1):
    row = "".join(chr(65 + j) for j in range(i))
    print(row)
```

### 5. Alphabet diamond

```python
n = 4

for i in range(1, n + 1):
    letters = "".join(chr(64 + j) for j in range(1, i + 1))
    print(" " * (n - i) + letters)

for i in range(n - 1, 0, -1):
    letters = "".join(chr(64 + j) for j in range(1, i + 1))
    print(" " * (n - i) + letters)
```

### 6. Letter with its ASCII value

```python
n = 5

for i in range(n):
    letter = chr(65 + i)
    print(f"{letter} -> {ord(letter)}")
```
