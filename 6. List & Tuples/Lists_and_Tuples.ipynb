# Lists and TuplesBoth **lists** and **tuples** store a collection of items in order. The big difference:- **List** → changeable (mutable), written with `[ ]`- **Tuple** → unchangeable (immutable), written with `( )`Think of a list as a shopping cart you keep editing, and a tuple as a receipt — once it's printed, it's fixed.

## Lists

```python
fruits = ["apple", "banana", "cherry"]print(fruits)print(fruits[0])      # indexingprint(fruits[-1])     # last itemprint(fruits[0:2])    # slicing
```

### Common List Methods| Method | What it does ||---|---|| `.append()` | add item to the end || `.insert()` | add item at a specific position || `.remove()` | remove item by value || `.pop()` | remove item by index (returns it) || `.sort()` | sort in place || `.reverse()` | reverse in place || `.extend()` | add multiple items from another list |

```python
fruits.append("orange")print(fruits)fruits.insert(1, "mango")print(fruits)fruits.remove("banana")print(fruits)popped = fruits.pop(0)print("Removed:", popped)print(fruits)fruits.sort()print(fruits)fruits.extend(["kiwi", "grape"])print(fruits)
```

Lists are **mutable** — you can change items directly.

```python
nums = [1, 2, 3]nums[0] = 100print(nums)
```

## Tuples

```python
point = (10, 20)colors = "red", "green", "blue"   # parentheses are optionalprint(point)print(colors)print(point[0])
```

Tuples are **immutable** — trying to change them raises an error.

```python
try:    point[0] = 99except TypeError as e:    print("Error:", e)
```

### Tuple UnpackingA neat trick — assign multiple variables in one line.

```python
dimensions = (1920, 1080)width, height = dimensionsprint(width)print(height)
```

## List vs Tuple — When to Use What| | List | Tuple ||---|---|---|| Syntax | `[ ]` | `( )` || Mutable? | Yes | No || Use when | Data will change | Data is fixed (coordinates, RGB values, etc.) || Performance | Slightly slower | Slightly faster |## Quick Recap- Lists are flexible and editable; tuples are locked once created.- Use `.append()`, `.remove()`, `.pop()`, `.sort()` to manage lists.- Use tuples for fixed groups of data and easy unpacking.
